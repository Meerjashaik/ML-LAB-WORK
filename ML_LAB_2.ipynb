```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Ali', 'Rahul', 'Priya'],
    'Marks': [85, np.nan, 95]
})

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ali', 'Rahul', 'Ali', 'Priya'],
    'Marks': [85, 90, 85, 95]
})

df = df.drop_duplicates()

print(df)
```
